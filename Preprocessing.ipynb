```python
### Step 1: Load the Dataset and Handle Missing Data
import pandas as pd

# Load dataset (assuming a CSV file named 'student_performance.csv')
df = pd.read_csv('college_student_placement_dataset.csv')
df.head()
```

```python
# Check for missing data
print(df.isnull().sum())
```

```python
# Strategies to handle missing data:
# 1. Replace missing data with mean, median, or mode.
# 2. Remove rows with missing data.

# Example: Replace missing numerical data with the mean
#df.fillna(df.mean(), inplace=True)
```

```python
# Example: For categorical data, replace missing data with the mode
for column in df.select_dtypes(include=['object']).columns:
    df.fillna(df[column].mode()[0], inplace=True)
```

```python
### Step 2: Display Statistical Information and Data Types: Display basic statistical information and check data types.

# Display statistical information
print(df.describe())
```

```python
# Display data types
print(df.dtypes)
```

```python
### Step 3: Create a New Column `average_score`

### Create a new column that calculates the mean of the three exam scores: G1, G2, and G3.

# Ensure G1, G2, G3 are numerical
df[['IQ ', 'Extra_Curricular_Score', 'Academic_Performance']] = df[['IQ', 'Extra_Curricular_Score', 'Academic_Performance']].apply(pd.to_numeric, errors='coerce')

# Create the average_score column
df['average_score'] = df[['IQ ', 'Extra_Curricular_Score', 'Academic_Performance']].mean(axis=1)

print(df[['IQ ', 'Extra_Curricular_Score', 'Academic_Performance', 'average_score']].head())
```

```python
### Step 4: Find Students Scoring Above 5 in G3 and Count by Mother's Job

###Find students who scored above 5 in G3 and display the count by their mother's job.

# Filter students with G3 > 5
above_five_g3 = df[df['Academic_Performance'] > 5]

# Count by Mother's job (Mjob)
Placement = above_five_g3['Placement'].value_counts()

print(Placement)
```
