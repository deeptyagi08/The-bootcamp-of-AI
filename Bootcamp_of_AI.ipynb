```python
#  Create a list of random numbers and print all the even numbers in the list
import random

numbers = [random.randint(1, 100) for _ in range(20)]

print("List of numbers:")
print(numbers)

print("Even numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num)
```

```python
#  Create a string and print it in reverse order. Now reverse the words also.
text = "Python is easy to learn"

print("Original string:")
print(text)

print("\nString in reverse order:")
print(text[::-1])

print("\nWords in reverse order:")
words = text.split()
print(" ".join(words[::-1]))
```
