```python
# Sample 2 dimensional data creation in Numpy (hint-random)
import numpy as np

# Generate random 2-dimensional data
data = np.random.rand(10, 2)

print("2D Random Data:")
print(data)

print("\nShape of data:")
print(data.shape)
```

```python
# Use pandas features for data cleaning (Hint-RegEx)
import pandas as pd

# Sample data
data = {
    "Name": [" Rahul ", "Priya@", "Amit123", " Sneha "],
    "Email": ["rahul@gmail.com", "priya@gmail.com", "amit@gmail.com", "sneha@gmail.com"]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Remove special characters from Name using RegEx
df["Name"] = df["Name"].str.replace(r"[^A-Za-z ]", "", regex=True)

# Remove leading and trailing spaces
df["Name"] = df["Name"].str.strip()

print("\nCleaned Data:")
print(df)
```

```python
# Use Matplotlib and draw a Sine wave
import numpy as np
import matplotlib.pyplot as plt

# Generate values
x = np.linspace(0, 2 * np.pi, 100)

# Calculate sine values
y = np.sin(x)

# Plot sine wave
plt.plot(x, y)

plt.title("Sine Wave")
plt.xlabel("X")
plt.ylabel("sin(X)")
plt.grid(True)

plt.show()
```
